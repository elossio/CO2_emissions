# Loading and Consolidation of CO2 Emission Datasets

This notebook loads several CSV files containing CO2 emission data from vehicles across different periods and consolidates them into a single database.

## 1. Imports and Configurations

We import the pandas library and the path constants defined in the project configuration file.

```python
import pandas as pd

from src.config import (
    DATA_2005_2014,
    DATA_2015_2019,
    DATA_2020,
    DATA_2021,
    DATA_2022,
    DATA_2023,
    DATA_2024,
    CONSOLIDATED_DATA,
)
```

## 2. Loading Data

We read the individual CSV files. Note that for the years 2023 and 2024, it is necessary to specify 'latin' encoding.

```python
df_2005_2014 = pd.read_csv(DATA_2005_2014)
df_2015_2019 = pd.read_csv(DATA_2015_2019)
df_2020 = pd.read_csv(DATA_2020)
df_2021 = pd.read_csv(DATA_2021)
df_2022 = pd.read_csv(DATA_2022)
df_2023 = pd.read_csv(DATA_2023, encoding="latin")
df_2024 = pd.read_csv(DATA_2024, encoding="latin")
```

## 4. Consolidação das Bases

Concatenamos todos os DataFrames em um único objeto e resetamos o índice para manter a continuidade.

```python
# Consolidando bases e resetando índice
df = pd.concat(
    [df_2005_2014, df_2015_2019, df_2020, df_2021, df_2022, df_2023, df_2024]
)

df = df.reset_index(drop=True)

df.head()
```

## 7. Padronização dos Nomes das Colunas

Aplicamos uma padronização snake_case aos nomes das colunas para facilitar o uso no desenvolvimento e análise.

```python
import re


def string_to_snake_case(string):
    string = re.sub("/", " ", string)                       # substitui / por espaço
    string = re.sub(" +", " ", string)                      # substitui múltiplos espaços por um espaço
    string = re.sub(" ", "_", string)                       # substitui espaço por _
    string = re.sub("([a-z0-9])([A-Z])", r"\1_\2", string)  # adiciona _ entre letras minúsculas e maiúsculas
    string = re.sub(r"\(", "", string)                      # elimina parênteses
    string = re.sub(r"\)", "", string)                      # elimina parênteses
    return string.lower()                                   # retorna a string em minúsculo
```

```python
[string_to_snake_case(coluna) for coluna in df.columns]
```

```python
df.columns = [string_to_snake_case(coluna) for coluna in df.columns]

df.head()
```
